#### Prepare and execute SQL to pull out required companies

```python
# prepare sql statement to pull out companies

like_conditions = " or ".join([f"nimi like ?" for _ in ettevõtte_nimed])
regkood_conditions = " or ".join([f"ariregistri_kood = ?" for _ in ettevõtte_regkoodid])
params = [f"{t}" for t in ettevõtte_nimed] + list(ettevõtte_regkoodid)

#list of companies from source
juriidilised_isikud = duckdb.sql(f"""
    select * from '{üld_andmed}'
    where {like_conditions} 
       or ({regkood_conditions})
""", params=params).df()

# for debugging
#tulemus=juriidilised_isikud[juriidilised_isikud["ariregistri_kood"]==90006845]
#tulemus
```

#### Get annual report header information

```python
# registry codes required for reporting header information
registrikoodid = juriidilised_isikud["ariregistri_kood"].astype(int).tolist()

# get reporting header information
aruande_päis = duckdb.sql(f"""
    select * from read_csv(
        '{maa_üld}',
        delim=';',
        header=true,
        types={{'Audiitorettevõtja (AEV)': 'VARCHAR'}}
    )
    where registrikood in (select unnest(?)) 
    and aruandeaasta in (select unnest(?))
""", params=[registrikoodid, aastad]).df()

# get list of report id-s
raportite_id = aruande_päis["report_id"].astype(int).tolist()
```

#### Get EMTAK information

```python
# result of emtak query
emtak_myygitulu = duckdb.sql(f"""
    select 
        j.ariregistri_kood 
        ,j.nimi
        ,ap.aruandeaasta
        ,emtak
        ,"Jaotatud müügitulu" as jaotatud_müügitulu
        ,põhitegevusala
        ,emtak_version
    from read_csv(
        '{emtak_andmed}',
        delim=';',
        header=true
    ) mt
    inner join aruande_päis ap on mt.report_id = ap.report_id
    inner join juriidilised_isikud j on ap.registrikood = j.ariregistri_kood
    where mt.report_id in (select unnest(?))
"""
, params=[raportite_id]).df()

```

### Get annual accounting figures

```python
# get annual accouting figures into maa variable
maa = duckdb.sql("""
    select
        j.ariregistri_kood 
        ,j.nimi
        ,ap.aruandeaasta
        ,maa.tabel as aruanne
        ,maa.elemendi_label as rida
        ,vaartus as väärtus
        --,* 
    from read_csv(
        ?,
        delim=';',
        header=true,
        union_by_name=true
    ) as maa
    inner join aruande_päis ap on maa.report_id = ap.report_id
    inner join juriidilised_isikud j on ap.registrikood = j.ariregistri_kood
    where maa.report_id in (select unnest(?))
"""
,params=[maa_failid, raportite_id]).df()
```
